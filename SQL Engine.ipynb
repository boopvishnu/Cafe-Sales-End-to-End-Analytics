```python
import pandas as pd
from sqlalchemy import create_engine

# Load Excel File
file_path = r"C:\Users\eichi\OneDrive\Desktop\Daily Guide\Dirty to clean.xlsx"
df = pd.read_excel(file_path)

# Fixing Spaces And Hyphens and Replace With Underscore
df.columns = df.columns.str.replace(' ', '_').str.replace('-', '_')

# MySQL Server connection engine
engine = create_engine("mysql+pymysql://root:101184@localhost/end_projects")

# Sending Data in Tables
df.to_sql('sales_data', con=engine, if_exists='append', index=False)

print("Data Succefully Loaded")
```
